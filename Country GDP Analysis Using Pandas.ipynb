#plot the graph using one variable is called uni variant analysis

```python
vis2=sns.displot(df['InternetUsers'])
plt.show(vis2)
```

```python
vis3=sns.distplot(df["InternetUsers"],bins=15)
plt.show(vis3)
```

```python
plt.rcParams['figure.figsize']=6,3
```

```python
vis4=sns.boxplot(data=df,x="IncomeGroup",y="BirthRate")
plt.show(vis4)
```

#plot the graph using two variable is called uni variant analysis

```python
vis5=sns.lmplot(data=df,x="InternetUsers",y="BirthRate")
plt.show(vis5)
```

```python
vis5=sns.lmplot(data=df,x="InternetUsers",y="BirthRate",fit_reg=False)
plt.show(vis5)
```

```python
vis5=sns.lmplot(data=df,x="InternetUsers",y="BirthRate",fit_reg=True)
plt.show(vis5)
```

```python
vis5=sns.lmplot(data=df,x="InternetUsers",y="BirthRate",fit_reg=False,hue='IncomeGroup')
plt.show(vis5)
```
